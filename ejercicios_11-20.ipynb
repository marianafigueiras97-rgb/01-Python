### EJERCICIO 11
Escribe un programa que pida al usuario que introduzca su edad. Si el usuario ingresa un valor no numérico o un valor fuera del rango esperado (por ejemplo, menor que 0 o mayor que 120), maneja las excepciones adecuadamente.

```python
class UnderAge(Exception):
    pass

class OverAge(Exception):
    pass

def age_request():
    user_age = int(input("Inserta tu edad"))

    if user_age < 0:
        raise UnderAge("la edad debe ser mayor a 0")
    elif user_age > 120:
        raise OverAge("la edad no debe ser mayor a 120")
    else:
        return print(f'la edad del usuario es {user_age}')

try:
    age_request()
except UnderAge as error:
    print(f'Error: {error}')
except OverAge as error:
    print(f'Error: {error}')
except ValueError as error:
    print(f'Error: {error}')
```

### EJERCICIO 12
Genera una función que, al recibir una frase, devuelva una lista con la longitud de cada palabra. Usa la función map().

```python
#pedir usuario frase
user_phrase = input("Introduce un frase")

def each_word_len(string):
    # separarlo con espacios y hacer lista de palabras 
    words = string.split(" ")
    # crear funcion auxiliar que obtiene un lista con las longitudes de cada
    lengths = list(map(len,words))
    return lengths

result = each_word_len(user_phrase)
print(result)
# input introducido hola oomo estas tu 
```

### EJERCICIO 13
Genera una función que, para un conjunto de caracteres, devuelva una lista de tuplas con cada letra en mayúsculas y minúsculas. Las letras no pueden estar repetidas. Usa la función map().

```python
caracteres = {'a','b','c','d','e','a'}
caracteres2 = {'a', 'k', 'm', 'z', 'q', 'h', 't'}
caracteres3 = {'B', 'F', 'L', 'X', 'R', 'P', 'N'}
# usamos sets para evitar que se repitan los caracteres

def get_caracter_min_mayus(conjunto):
    # funcion auxiliar para convertir el caracter a mayuscula y minuscula
    convert = lambda letra: (letra.upper(), letra.lower())
    # map que por cada caracter del conjunto ejecuta la funcion auxiliar. Usamos list para que nos lo devuelva en una lista
    tuplas = list(map(convert,conjunto))
    return tuplas

# ejecucion con 3 ejemplo 
print(get_caracter_min_mayus(caracteres))
print(get_caracter_min_mayus(caracteres2))
print(get_caracter_min_mayus(caracteres3))
```

### EJERCICIO 14
Crea una función que retorne las palabras de una lista que comiencen con una letra en específico. Usa la función filter().

```python
lista_palabras = ['avion','burro','cabeza','dado','avioneta','balcon','cera','dardo']

def comienza_por(palabra , letra):
    if palabra[0] == letra:
        return True
    return False

def filtrado_palabras(lista,letra):
    result = list(filter(lambda palabra: comienza_por(palabra,letra),lista))
    return result


print(filtrado_palabras(lista_palabras,'a'))
print(filtrado_palabras(lista_palabras,'b'))
print(filtrado_palabras(lista_palabras,'c'))
print(filtrado_palabras(lista_palabras,'d'))
```
