### EJERCICIO 10
Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una excepción personalizada y maneja el error adecuadamente.

```python
# definimos la excepcion personalizada
class EmptyList(Exception):
    pass

# funcion con su bloque try except
def average(list):
    try:
        if not list:
            # Lanza el error si está vacía
            raise EmptyList("No se puede operar con una lista vacía.")
        
        # Si no esta vacia
        return sum(list) / len(list)
    except EmptyList as error:
        # Captura e imprime el error por consola 
        print(f"Error detectado: {error}")

# EJEMPLO DE USO 
#lista vacía
lista_de_prueba = []
average(lista_de_prueba)

#lista llena
lista_de_prueba2 = [1,2,3,4]
print(average(lista_de_prueba2))

#lista llena otro
lista_de_prueba3 = [1.0,20,5.32,4]
print(average(lista_de_prueba3))
```

### EJERCICIO 11
Escribe un programa que pida al usuario que introduzca su edad. Si el usuario ingresa un valor no numérico o un valor fuera del rango esperado (por ejemplo, menor que 0 o mayor que 120), maneja las excepciones adecuadamente.

```python
class UnderAge(Exception):
    pass

class OverAge(Exception):
    pass

def age_request():
    user_age = int(input("Inserta tu edad"))

    if user_age < 0:
        raise UnderAge("la edad debe ser mayor a 0")
    elif user_age > 120:
        raise OverAge("la edad no debe ser mayor a 120")
    else:
        return print(f'la edad del usuario es {user_age}')

try:
    age_request()
except UnderAge as error:
    print(f'Error: {error}')
except OverAge as error:
    print(f'Error: {error}')
except ValueError as error:
    print(f'Error: {error}')
```
