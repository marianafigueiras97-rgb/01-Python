### EJERCICIO 1 
Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena.
Los espacios no deben ser considerados.

```python
def letter_frecuency(text):
    """
    Recibe una cadena de texto
    La normaliza bajando a minúsculas y removiendo espacios
    La convierte a una lista que se recorre
    Va añadiendo cada letra commo atributo del diccionario 
    Asigna un valor a cada letra usando count()
    Imprime por consola el resultado

    Args:
        text (string): texto a evaluar
    """
    lista = list(text.lower().replace(" ",""))

    dicc = {}
    for letter in lista:
        value = lista.count(letter)
        dicc[letter] = value

    print(dicc)

letter_frecuency('Mariana')
letter_frecuency('José')
letter_frecuency('Quiero ganar mucho dinero')
```

### EJERCICIO 2 
Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().

```python
numbers = [16,4,25,7,14,9]
double = lambda num1: num1 * 2
double_numbers = map(double,numbers)
print(list(double_numbers))
```
