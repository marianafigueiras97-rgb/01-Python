### EJERCICIO 22
Dada una lista numérica, obtén el producto total de los valores. Usa la función `reduce()`.

```python
from functools import reduce 

lista = [3,3,3]
lista2 = [1,5,7]
lista3 = [3,4,9,5]
lista4 = [20,10,5]

def producto_de_lista(listado):
    producto =  reduce(lambda num1, num2: num1 * num2,listado)
    print(producto)
    
producto_de_lista(lista)
producto_de_lista(lista2)
producto_de_lista(lista3)
producto_de_lista(lista4)
```

### EJERCICIO 23
Concatena una lista de palabras. Usa la función `reduce()`.

```python
from functools import reduce 
lista_ejemplo = ['hola','como','estas']
naturaleza = ['sol', 'viento', 'mar', 'tierra', 'fuego', 'bosque', 'rio']

def concat_reduce (lista_palabras):
    concat = reduce(lambda a,b: f"{a} {b}" ,lista_palabras)
    return concat

print(concat_reduce(lista_ejemplo))
print(f"Los siguientes elementos pertenecen a la naturaleza: {concat_reduce(naturaleza)}")
```

### EJERCICIO 24
Calcula la diferencia total en los valores de una lista. Usa la función `reduce()`.

### EJERCICIO 28
Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.

```python
lista_con_duplicados = [1,2,3,4,5,1]
lista_con_duplicados2 = ['hola','como','hola','estas']


def find_duplicates(lista):
    vistos = set()
    duplicados = set()
    
    for item in lista:
        if item in vistos:
            duplicados.add(item)
        else:
            vistos.add(item)
    
    return list(duplicados)

print(find_duplicates(lista_con_duplicados))
find_duplicates(lista_con_duplicados2)
```

### EJERCICIO 29
Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter `#` excepto los últimos cuatro.

```python
ejemplo_num = 1546879
ejemplo_texto = "ajhUl*4se2"
ejemplo_pequeño = "ABC"

def enmascarar(variable):
    cadena = str(variable)
    vueltas = len(cadena) - 4
    ac = 0
    for item in cadena:
        if ac < vueltas:
            cadena = cadena.replace(item,"#")
            ac += 1
    print(cadena)
    
enmascarar(ejemplo_num)
enmascarar(ejemplo_texto)
enmascarar(ejemplo_pequeño)
```

### EJERCICIO 30
Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.

```python
palabra1 ='amor'
palabra2 ='roma'
palabra3 ='peso'
palabra4 ='pose'
palabra5 ='hola'
palabra6 ='chao'

def is_anagrama(palabra1,palabra2):
    for letra in palabra1:
        if letra in palabra2:
            continue
        else:
            return False
    return True

print(is_anagrama(palabra1,palabra2))
print(is_anagrama(palabra3,palabra4))
print(is_anagrama(palabra5,palabra6))
```
