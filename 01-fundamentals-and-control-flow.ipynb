### EJERCICIO 37
Genera un programa que nos indique si es de noche, de día o de tarde según la hora proporcionada por el usuario.

```python
from datetime import datetime, time 

inicio_madrugada = time(0,0)
fin_madrugada = time(5,59)
inicio_dia = time(6,0)
fin_dia = time(11,59)
inicio_tarde = time(12,0)
fin_tarde = time(19,59)
inicio_noche = time(20,00)
fin_noche = time(23,59)


def indicar_momento_del_dia():
    hora_introducida = input("Introduce la hora en formato HH:MM")
    try:
        hora_usuario = datetime.strptime(hora_introducida,"%H:%M").time()
        print(f"hora recibida : {hora_usuario}")
            
        if inicio_madrugada <= hora_usuario <= fin_madrugada:
            print("Es de madrugada.")
        elif inicio_dia <= hora_usuario <= fin_dia:
            print("Es de día.")
        elif inicio_tarde <= hora_usuario <= fin_tarde:
            print("Es de tarde.")
        else:
            print("Es de noche.")
            
    except ValueError:
        print("formato de hora incorrecto, intente de nuevo")
        

for i in range(1,6):
    indicar_momento_del_dia()
```

### EJERCICIO 39
Escribe una función que tome dos parámetros:
- `figura` (una cadena que puede ser `"rectangulo"`, `"circulo"` o `"triangulo"`).
- `datos` (una tupla con los datos necesarios para calcular el área de la figura).

```python
import math 

def calcular_area_figura(figura,datos):
    if figura == "rectangulo":
        area = datos[0] * datos[1]
        print(f"el area del {figura} es {area} cm²")
    elif figura == "circulo":
        area = (math.pi) * (datos ** 2) 
        print(f"el area del {figura} es {area} cm²")
    else:
        area = (datos[0] * datos[1]) /2
        print(f"el area del {figura} es {area} cm²")
    
calcular_area_figura("rectangulo",(2,4))
calcular_area_figura("circulo",(4))
calcular_area_figura("triangulo",(2,4))
```
