 ### EJERCICIO 6
Escribe una función que calcule el factorial de un número de manera recursiva.

```python
def factorial(numero):
    if numero <= 1:
        return 1
    else:
        v = factorial (numero - 1)
        result = numero * v
        return result

print(factorial(5))
print(factorial(4))
print(factorial(3))
print(factorial(2))
print(factorial(1))
```

Crea una función que calcule el promedio de una lista de números.

```python
from functools import reduce


lista_numeros = [1,4,7,5,3,2]
lista_numeros2 = [10,10,10]
lista_numeros3 = [20,14,16,5]

def average(lista):
    total = reduce(lambda a,b: a + b,lista)
    average = total / len(lista)
    print(average)
    
average(lista_numeros2)
```

### EJERCICIO 28
Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.

```python
lista_con_duplicados = [1,2,3,4,5,1]
lista_con_duplicados2 = ['hola','como','hola','estas']


def find_duplicates(lista):
    vistos = set()
    duplicados = set()
    
    for item in lista:
        if item in vistos:
            duplicados.add(item)
        else:
            vistos.add(item)
    
    return list(duplicados)

print(find_duplicates(lista_con_duplicados))
find_duplicates(lista_con_duplicados2)
```

### EJERCICIO 29
Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter `#` excepto los últimos cuatro.

```python
ejemplo_num = 1546879
ejemplo_texto = "ajhUl*4se2"
ejemplo_pequeño = "ABC"

def enmascarar(variable):
    cadena = str(variable)
    vueltas = len(cadena) - 4
    ac = 0
    for item in cadena:
        if ac < vueltas:
            cadena = cadena.replace(item,"#")
            ac += 1
    print(cadena)
    
enmascarar(ejemplo_num)
enmascarar(ejemplo_texto)
enmascarar(ejemplo_pequeño)
```

### EJERCICIO 30
Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.

```python
palabra1 ='amor'
palabra2 ='roma'
palabra3 ='peso'
palabra4 ='pose'
palabra5 ='hola'
palabra6 ='chao'

def is_anagrama(palabra1,palabra2):
    for letra in palabra1:
        if letra in palabra2:
            continue
        else:
            return False
    return True

print(is_anagrama(palabra1,palabra2))
print(is_anagrama(palabra3,palabra4))
print(is_anagrama(palabra5,palabra6))
```

### EJERCICIO 32
Crea una función que tome un nombre completo y una lista de empleados, busque el nombre en la lista y devuelva el puesto del empleado si se encuentra; de lo contrario, devuelve un mensaje indicando que la persona no trabaja aquí.

```python
empleados = [
    {"nombre": "Ana García", "puesto": "Desarrolladora"},
    {"nombre": "Pedro López", "puesto": "Diseñador"},
    {"nombre": "Laura Pérez", "puesto": "Project Manager"},
    {"nombre": "Carlos Díaz", "puesto": "Analista"}
]

def buscar_puesto_empleado(nombre_empleado):
    encontrado = False
    for empleado in empleados:
        if nombre_empleado == empleado["nombre"]:
            encontrado = True
    
    if encontrado == True:
        print(f'el puesto de {nombre_empleado} es {empleado["puesto"]}')
    else:
        print("Esta persona no trabaja aquí")
            
nombre_a_buscar = input("Introduce el nombre del empleado")
buscar_puesto_empleado(nombre_a_buscar)
```
