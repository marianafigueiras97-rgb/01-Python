### EJERCICIO 31
Crea una función que solicite al usuario ingresar una lista de nombres y luego un nombre para buscar en esa lista. Si el nombre está en la lista, imprime un mensaje indicando que fue encontrado; de lo contrario, lanza una excepción.

```python
class NotFound(Exception):
    pass

lista_usuario = []

def rellenar_Lista():
    print("vamos a añadir 4 nombres a una lista")
    while len(lista_usuario) < 4:
        lista_usuario.append(input("añade un nombre a la lista"))
    print(f'la lista final es {lista_usuario}')

def buscar_nombre(lista):
    print("Ahora escribe un nombre que quieras guardar en la lista")
    nombre_a_buscar = input("escribe el nombre")
    print(f'el nombre a buscar es {nombre_a_buscar}')
    if nombre_a_buscar in lista:
        print(f'Encontrado {nombre_a_buscar} esta en la lista')
    else:
        raise NotFound(f'{nombre_a_buscar} no se encuentra en la lista')

try:
    rellenar_Lista()
    buscar_nombre(lista_usuario)
except NotFound as error:
    print(f'Error: {error}')
    
# OUTPUT ENCONTRADO
# vamos a añadir 4 nombres a una lista
# la lista final es ['mariana', 'manuel', 'trina', 'isabel']
# Ahora escribe un nombre que quieras guardar en la lista
# el nombre a buscar es mariana
# Encontrado mariana esta en la lista
```

### EJERCICIO 32
Crea una función que tome un nombre completo y una lista de empleados, busque el nombre en la lista y devuelva el puesto del empleado si se encuentra; de lo contrario, devuelve un mensaje indicando que la persona no trabaja aquí.

```python
empleados = [
    {"nombre": "Ana García", "puesto": "Desarrolladora"},
    {"nombre": "Pedro López", "puesto": "Diseñador"},
    {"nombre": "Laura Pérez", "puesto": "Project Manager"},
    {"nombre": "Carlos Díaz", "puesto": "Analista"}
]

def buscar_puesto_empleado(nombre_empleado):
    encontrado = False
    for empleado in empleados:
        if nombre_empleado == empleado["nombre"]:
            encontrado = True
    
    if encontrado == True:
        print(f'el puesto de {nombre_empleado} es {empleado["puesto"]}')
    else:
        print("Esta persona no trabaja aquí")
            
nombre_a_buscar = input("Introduce el nombre del empleado")
buscar_puesto_empleado(nombre_a_buscar)
```

### EJERCICIO 34
Crea la clase `Arbol`.

Define un árbol genérico con un tronco y ramas como atributos.

**Métodos disponibles:**
- `crecer_tronco`
- `nueva_rama`
- `crecer_ramas`
- `quitar_rama`
- `info_arbol`

**Código a seguir:**
- Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.
- Implementar el método `crecer_tronco` para aumentar la longitud del tronco en una unidad.
- Implementar el método `nueva_rama` para agregar una nueva rama de longitud 1 a la lista de ramas.
- Implementar el método `crecer_ramas` para aumentar en una unidad la longitud de todas las ramas existentes.
- Implementar el método `quitar_rama` para eliminar una rama en una posición específica.
- Implementar el método `info_arbol` para devolver información sobre la longitud del tronco, el número de ramas y sus longitudes.

**Caso de uso:**
1. Crear un árbol.
2. Hacer crecer el tronco una unidad.
3. Añadir una nueva rama.
4. Hacer crecer todas las ramas una unidad.
5. Añadir dos nuevas ramas.
6. Retirar la rama situada en la posición 2.
7. Obtener información sobre el árbol.

```python
class Arbol:
    
    def __init__(self):
        # constructor del arbol con sus atributos
        # logitud inicial del arbol
        self.tronco = 1
        # listado de ramas vacío
        self.ramas = []
    
    def crecer_tronco(self):
        self.tronco = self.tronco + 1
        print("arbol creció con exito!✅")
    
    def nueva_rama(self):
        # añadir una nueva rama a la lista
        self.ramas.append(1)
        print("nueva rama de longitud 1 añadida con exito!✅")
    
    def crecer_ramas(self):
        # con un bucle recorrer la lista de ramas y sumarle 1 
        for i in range(len(self.ramas)):
            self.ramas[i] += 1
        print("todas las ramas han crecido en 1✅")
        
    
    def quitar_rama(self,posicion):
        # remover la rama de la lista
        if len(self.ramas) > posicion:
            self.ramas.remove(posicion)
            print(f"rama en posicion {posicion} eliminada con exito ✅")
            
        else:
            print("Error ❌: no se puede eliminar esa rama porque no existe")
    
    def info_arbol(self):
        # imprimir longitud de arbol, numero de ramas y longitud de cada una de ellas
        print(f'La longitud del tronco es {self.tronco} tiene {len(self.ramas)} ramas y sus longitudes son {self.ramas}')
    

# 1. Crear un árbol.
arbol_1 = Arbol()
arbol_1.info_arbol()

# 2. Hacer crecer el tronco una unidad.
arbol_1.crecer_tronco()

# 3. Añadir una nueva rama.
arbol_1.nueva_rama()

# 4. Hacer crecer todas las ramas una unidad.
arbol_1.crecer_ramas()

# 5. Añadir dos nuevas ramas.
arbol_1.nueva_rama()
arbol_1.nueva_rama()

# 6. Retirar la rama situada en la posición 2.
arbol_1.quitar_rama(2)
# 7. Obtener información sobre el árbol.
arbol_1.info_arbol()
```
