### EJERCICIO 17
Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, `[5,7,2]` corresponde al número `572`. Usa la función `reduce()`.

```python
from functools import reduce 

lista_digitos = [1,2,3]
lista_digitos2 = [4,8,1,2,4]
lista_digitos3 = [1,9,9,7]

def convertir_a_digito(lista):
    # definimos la funcion que se le pasara como argumento a filter 
    # lo que hace es coger el acumulador multiplicarlo por 10 y sumarle el siguiente digito 
    digito = reduce(lambda ac, digito: ac * 10 + digito,lista)
    return digito


print(convertir_a_digito(lista_digitos))
print(convertir_a_digito(lista_digitos2))
print(convertir_a_digito(lista_digitos3))
```

### EJERCICIO 22
Dada una lista numérica, obtén el producto total de los valores. Usa la función `reduce()`.

```python
from functools import reduce 

lista = [3,3,3]
lista2 = [1,5,7]
lista3 = [3,4,9,5]
lista4 = [20,10,5]

def producto_de_lista(listado):
    producto =  reduce(lambda num1, num2: num1 * num2,listado)
    print(producto)
    
producto_de_lista(lista)
producto_de_lista(lista2)
producto_de_lista(lista3)
producto_de_lista(lista4)
```

### EJERCICIO 23
Concatena una lista de palabras. Usa la función `reduce()`.

```python
from functools import reduce 
lista_ejemplo = ['hola','como','estas']
naturaleza = ['sol', 'viento', 'mar', 'tierra', 'fuego', 'bosque', 'rio']

def concat_reduce (lista_palabras):
    concat = reduce(lambda a,b: f"{a} {b}" ,lista_palabras)
    return concat

print(concat_reduce(lista_ejemplo))
print(f"Los siguientes elementos pertenecen a la naturaleza: {concat_reduce(naturaleza)}")
```

### EJERCICIO 24
Calcula la diferencia total en los valores de una lista. Usa la función `reduce()`.

```python
from functools import reduce 

lista = [3,3,3]
lista2 = [1,5,7]
lista3 = [3,4,9,5]
lista4 = [20,10,5]

def diferencia_de_lista(listado):
    diferencia =  reduce(lambda num1, num2: num1 - num2,listado)
    print(diferencia)
    
diferencia_de_lista(lista)
diferencia_de_lista(lista2)
diferencia_de_lista(lista3)
diferencia_de_lista(lista4)
```
