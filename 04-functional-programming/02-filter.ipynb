### EJERCICIO 3
Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

```python
palabras_1 = [
    "casa", "casita", "caserío", "perro", "casco",
    "camión", "cascada", "mesa", "casamiento"
]
# cas
palabras_2 = ["programación", "programa", "programador",
    "python", "java", "diagrama", "telegrama"]
# gram
palabras_3 = [
    "gato", "gata", "gatito", "perro",
    "conejo", "lagarto", "gatuna"
]
# gat

def search (list,text):
    """
    Búsca coincidencias de texto y devuelve una lista con las palabras encontradas
    Uso de list comprehension con condicional para ver si hay match

    Args:
        list (list): lista original
        text (string): objetivo de búsqueda
    """
    matches = [word for word in list if text in word]
    print(matches) 

search(palabras_1,'cas')
search(palabras_2,'gram')
search(palabras_3,'gat')
```

## EJERCICIO 9
Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista excluyendo ciertas mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]. Usa la función filter().

```python
# listas de ejemplo
mascotas = ["Perro", "Gato", "Conejo", "Mapache","Canario", "Hámster", "Tigre"]
mascotas2 = ["Perro", "Gato", "Conejo","Tortuga", "Loro", "Hurón"]
mascotas3 = ["Mapache", "Tigre","Serpiente Pitón", "Cocodrilo", "Oso","Gato"]

# funcion auxiliar que devuelve un booleano para pasarsela al filter
def pet_is_allowed(pet):
    prohibidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]
    return pet not in prohibidas

# funcion de filtrado
def pet_filter(pets_list):
    return list(filter(pet_is_allowed,pets_list))

# ejecucion con los 3 ejemplo 
print(pet_filter(mascotas))
print(pet_filter(mascotas2))
print(pet_filter(mascotas3))
```

### EJERCICIO 14
Crea una función que retorne las palabras de una lista que comiencen con una letra en específico. Usa la función filter().

```python
lista_palabras = ['avion','burro','cabeza','dado','avioneta','balcon','cera','dardo']

def comienza_por(palabra , letra):
    if palabra[0] == letra:
        return True
    return False

def filtrado_palabras(lista,letra):
    result = list(filter(lambda palabra: comienza_por(palabra,letra),lista))
    return result


print(filtrado_palabras(lista_palabras,'a'))
print(filtrado_palabras(lista_palabras,'b'))
print(filtrado_palabras(lista_palabras,'c'))
print(filtrado_palabras(lista_palabras,'d'))
```

### EJERCICIO 16
Escribe una función que tome una cadena de texto y un número entero `n` como parámetros y devuelva una lista de todas las palabras que sean más largas que `n`. Usa la función `filter()`.

```python
cadena_texto = "Esto es un ejemplo de cadena de texto"

def funcion_aux(palabra,numero):
    if len(palabra) > numero:
        return True
    return False

def mas_larga(cadena_texto,numero):
    # pasamos la cadena a una lista de palabras 
    listado = cadena_texto.split(" ")
    listado_filtrado = list(filter(lambda palabra: funcion_aux(palabra,numero),listado))
    print(listado_filtrado)

mas_larga(cadena_texto,3)
mas_larga(cadena_texto,4)
mas_larga(cadena_texto,5)
```

### EJERCICIO 18
Escribe un programa en Python que cree una lista de diccionarios con información de estudiantes (nombre, edad, calificación) y use `filter()` para extraer a los estudiantes con una calificación mayor o igual a 90.

```python
lista_estudiantes = [
    {
        "nombre": 'ana',
        "edad": 20,
        "nota":50
    },
    {
        "nombre": 'manuel',
        "edad": 25,
        "nota":99
    },
    {
        "nombre": 'trina',
        "edad": 45,
        "nota":95
    },
    {
        "nombre": 'paco',
        "edad": 30,
        "nota":89
    }
]
def is90plus(estudiante):
    if estudiante["nota"] >= 90:
        return True
    return False

def mejores_calificaciones(lista):
    
    lista_filtrada = list(filter(lambda estudiante: is90plus(estudiante),lista))
    return lista_filtrada

print(mejores_calificaciones(lista_estudiantes))
```

### EJERCICIO 20
Para una lista con elementos de tipo `integer` y `string`, obtén una nueva lista solo con los valores `int`. Usa la función `filter()`.

```python
lista_mezclada = ['hola',1,'gato',2,'perro',3,4,5,'chocolate','stitch']

def is_type_int(item):
    if type(item) is int:
        return True
    return False
    
lista_mezclada_filtrada = list(filter(is_type_int,lista_mezclada))
print(lista_mezclada_filtrada)
```
