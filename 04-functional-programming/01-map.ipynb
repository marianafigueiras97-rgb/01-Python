### EJERCICIO 2 
Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().

```python
numbers = [16,4,25,7,14,9]
double = lambda num1: num1 * 2
double_numbers = map(double,numbers)
print(list(double_numbers))
```

### EJERCICIO 4 
Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map().

```python
# Caso 1
lista1 = [10, 20, 30, 40]
lista2 = [1, 2, 3, 4]

# Caso 2
lista3 = [5, 8, 12, 15]
lista4 = [2, 4, 6, 8]

# Caso 3 
lista5 = [1, 2, 3, 4]
lista6 = [5, 4, 3, 2]

def calcular_diferencia (list1,list2):
    return list(map(lambda num1, num2: num1 - num2 ,list1,list2))

print(calcular_diferencia(lista1,lista2))
print(calcular_diferencia(lista3,lista4))
print(calcular_diferencia(lista5,lista6))
```

### EJERCICIO 7
Genera una función que convierta una lista de tuplas a una lista de strings. Usa la función map().

```python
personas = [
    ("María", 29),
    ("Carlos", 35),
    ("Ana", 22),
    ("Luis", 41),
    ("Sofía", 27)
]

estudiantes = [
    ("María", 8.5),
    ("Carlos", 4.7),
    ("Ana", 9.8),
    ("Luis", 6.2),
    ("Sofía", 5.0),
    ("Pedro", 3.9)
]

productos = [
    ("Leche", 1.35),
    ("Pan", 0.95),
    ("Huevos", 2.80),
    ("Arroz", 1.50),
    ("Pasta", 1.20)
]

def convertir(lista):
    texto = list(map(lambda item: str(item),lista))
    return texto

print(convertir(personas))
print(convertir(productos))
print(convertir(estudiantes))
```

### EJERCICIO 12
Genera una función que, al recibir una frase, devuelva una lista con la longitud de cada palabra. Usa la función map().

```python
#pedir usuario frase
user_phrase = input("Introduce un frase")

def each_word_len(string):
    # separarlo con espacios y hacer lista de palabras 
    words = string.split(" ")
    # crear funcion auxiliar que obtiene un lista con las longitudes de cada
    lengths = list(map(len,words))
    return lengths

result = each_word_len(user_phrase)
print(result)
# input introducido hola como estas tu 
```

### EJERCICIO 13
Genera una función que, para un conjunto de caracteres, devuelva una lista de tuplas con cada letra en mayúsculas y minúsculas. Las letras no pueden estar repetidas. Usa la función map().

```python
caracteres = {'a','b','c','d','e','a'}
caracteres2 = {'a', 'k', 'm', 'z', 'q', 'h', 't'}
caracteres3 = {'B', 'F', 'L', 'X', 'R', 'P', 'N'}
# usamos sets para evitar que se repitan los caracteres

def get_caracter_min_mayus(conjunto):
    # funcion auxiliar para convertir el caracter a mayuscula y minuscula
    convert = lambda letra: (letra.upper(), letra.lower())
    # map que por cada caracter del conjunto ejecuta la funcion auxiliar. Usamos list para que nos lo devuelva en una lista
    tuplas = list(map(convert,conjunto))
    return tuplas

# ejecucion con 3 ejemplo 
print(get_caracter_min_mayus(caracteres))
print(get_caracter_min_mayus(caracteres2))
print(get_caracter_min_mayus(caracteres3))
```
