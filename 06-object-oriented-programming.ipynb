### EJERCICIO 34
Crea la clase `Arbol`.

Define un árbol genérico con un tronco y ramas como atributos.

**Métodos disponibles:**
- `crecer_tronco`
- `nueva_rama`
- `crecer_ramas`
- `quitar_rama`
- `info_arbol`

**Código a seguir:**
- Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.
- Implementar el método `crecer_tronco` para aumentar la longitud del tronco en una unidad.
- Implementar el método `nueva_rama` para agregar una nueva rama de longitud 1 a la lista de ramas.
- Implementar el método `crecer_ramas` para aumentar en una unidad la longitud de todas las ramas existentes.
- Implementar el método `quitar_rama` para eliminar una rama en una posición específica.
- Implementar el método `info_arbol` para devolver información sobre la longitud del tronco, el número de ramas y sus longitudes.

**Caso de uso:**
1. Crear un árbol.
2. Hacer crecer el tronco una unidad.
3. Añadir una nueva rama.
4. Hacer crecer todas las ramas una unidad.
5. Añadir dos nuevas ramas.
6. Retirar la rama situada en la posición 2.
7. Obtener información sobre el árbol.

```python
class Arbol:
    
    def __init__(self):
        # constructor del arbol con sus atributos
        # logitud inicial del arbol
        self.tronco = 1
        # listado de ramas vacío
        self.ramas = []
    
    def crecer_tronco(self):
        self.tronco = self.tronco + 1
        print("arbol creció con exito!✅")
    
    def nueva_rama(self):
        # añadir una nueva rama a la lista
        self.ramas.append(1)
        print("nueva rama de longitud 1 añadida con exito!✅")
    
    def crecer_ramas(self):
        # con un bucle recorrer la lista de ramas y sumarle 1 
        for i in range(len(self.ramas)):
            self.ramas[i] += 1
        print("todas las ramas han crecido en 1✅")
        
    
    def quitar_rama(self,posicion):
        # remover la rama de la lista
        if len(self.ramas) > posicion:
            self.ramas.remove(posicion)
            print(f"rama en posicion {posicion} eliminada con exito ✅")
            
        else:
            print("Error ❌: no se puede eliminar esa rama porque no existe")
    
    def info_arbol(self):
        # imprimir longitud de arbol, numero de ramas y longitud de cada una de ellas
        print(f'La longitud del tronco es {self.tronco} tiene {len(self.ramas)} ramas y sus longitudes son {self.ramas}')
    

# 1. Crear un árbol.
arbol_1 = Arbol()
arbol_1.info_arbol()

# 2. Hacer crecer el tronco una unidad.
arbol_1.crecer_tronco()

# 3. Añadir una nueva rama.
arbol_1.nueva_rama()

# 4. Hacer crecer todas las ramas una unidad.
arbol_1.crecer_ramas()

# 5. Añadir dos nuevas ramas.
arbol_1.nueva_rama()
arbol_1.nueva_rama()

# 6. Retirar la rama situada en la posición 2.
arbol_1.quitar_rama(2)
# 7. Obtener información sobre el árbol.
arbol_1.info_arbol()
```

### EJERCICIO 35
Crea la clase `UsuarioBanco`.

Representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta corriente.

**Métodos:**
- `retirar_dinero`
- `transferir_dinero`
- `agregar_dinero`

**Código a seguir:**
- Inicializar un usuario con nombre, saldo y un indicador (`True` o `False`) de cuenta corriente.
- Implementar `retirar_dinero` para sustraer dinero del saldo, lanzando un error si no es posible.
- Implementar `transferir_dinero` para transferir dinero desde otro usuario, lanzando un error en caso de fallo.
- Implementar `agregar_dinero` para aumentar el saldo del usuario.

**Caso de uso:**
1. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.
2. Agregar 20 unidades al saldo de Bob.
3. Transferir 80 unidades de Bob a Alicia.
4. Retirar 50 unidades del saldo de Alicia.

```python
class UsuarioBanco:

    def __init__(self,nombre,saldo,cuenta_corriente):
        # constructor del usuario con sus atributos
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente

    def retirar_dinero(self,cantidad):
        if self.saldo > cantidad:
            self.saldo -= cantidad
            print(f"✅ {cantidad} € retirados exitosamente.")
        else:
            print("❌ Saldo insuficiente")


    def agregar_dinero(self,cantidad):
        self.saldo += cantidad
        print(f"{cantidad} € añadidos correctamente")

    def transferir_dinero(self, user, cantidad):
        if self.saldo > cantidad:
            self.saldo -= cantidad
            user.saldo += cantidad
        else:
            print("❌ Saldo insuficiente")

# 1. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.
alicia = UsuarioBanco("Alicia",100,True)
bob = UsuarioBanco("Bob",50,True)

# 2. Agregar 20 unidades al saldo de Bob.
bob.agregar_dinero(20)

# 3. Transferir 80 unidades de Bob a Alicia.
bob.transferir_dinero("Alicia",80)

# 4. Retirar 50 unidades del saldo de Alicia.
alicia.retirar_dinero(50)
```
